## 1. Importación limpia

```python
import pandas as pd

df_plataforma = pd.read_csv('estudiantes_cursos.csv', index_col=False)
print(df_plataforma.shape)
df_plataforma.head()
```

## 2. Serie y subconjunto

```python
serie_puntaje = df_plataforma['Puntaje_Evaluacion']
serie_puntaje.head()
```

```python
subconjunto = df_plataforma[['Curso', 'Modalidad', 'Lecciones_Completadas']]
subconjunto.head()
```

## 3. Registro puntual con .loc

```python
df_plataforma.loc[120]
```

## 4. Eliminar duplicados

```python
print('Duplicados:', df_plataforma.duplicated().sum())
df_plataforma = df_plataforma.drop_duplicates()
print('Filas restantes:', len(df_plataforma))
```

## 5. Filtro booleano (> 40 lecciones)

```python
df_plataforma[df_plataforma['Lecciones_Completadas'] > 40].head()
```

## 6. Filtro con .isin()

```python
df_plataforma[df_plataforma['Curso'].isin(['Python para Datos', 'Arquitectura Cloud'])].head()
```

## 7. Filtro con .between()

```python
df_plataforma[df_plataforma['Puntaje_Evaluacion'].between(70.0, 85.0)].head()
```

## 8. Consulta con .query()

```python
df_plataforma.query("Modalidad == 'Asincrónico' and Horas_Estudio > 25").head()
```

## 9. Filtro compuesto con .str.startswith()

```python
df_plataforma[
    (df_plataforma['Nivel_Educativo'] == 'Profesional') &
    (df_plataforma['Curso'].str.startswith('B'))
].head()
```

## 10. Imputación de nulos

```python
print(df_plataforma[['Horas_Estudio', 'Puntaje_Evaluacion']].isna().sum())

df_plataforma['Horas_Estudio'] = df_plataforma['Horas_Estudio'].fillna(df_plataforma['Horas_Estudio'].mean())
df_plataforma['Puntaje_Evaluacion'] = df_plataforma['Puntaje_Evaluacion'].fillna(0.0)

print(df_plataforma[['Horas_Estudio', 'Puntaje_Evaluacion']].isna().sum())
```

## 11. Transformación con .apply()

```python
def clasificar(nota):
    return 'Aprobado' if nota >= 70.0 else 'Reprobado'

df_plataforma['Estado_Materia'] = df_plataforma['Puntaje_Evaluacion'].apply(clasificar)
df_plataforma['Estado_Materia'].value_counts()
```

## 12. Renombrado y reordenamiento

```python
df_plataforma = df_plataforma.rename(columns={'Modalidad': 'Formato_Clase', 'Certificado_Generado': 'Graduado'})

primeras = ['Estado_Materia', 'ID_Estudiante']
resto = [c for c in df_plataforma.columns if c not in primeras]
df_plataforma = df_plataforma[primeras + resto]
df_plataforma.head()
```

## 13. Agrupación por Curso con .agg()

```python
df_plataforma.groupby('Curso').agg(
    Promedio_Horas=('Horas_Estudio', 'mean'),
    Total_Lecciones=('Lecciones_Completadas', 'sum')
)
```

## 14. Función personalizada `rango`

```python
def rango(series):
    return series.max() - series.min()

df_plataforma.groupby('Nivel_Educativo').agg(Rango_Puntaje=('Puntaje_Evaluacion', rango))
```

## 15. Agrupamiento multinivel

```python
df_plataforma.groupby(['Curso', 'Formato_Clase']).agg(
    Suma_Horas=('Horas_Estudio', 'sum'),
    Promedio_Puntaje=('Puntaje_Evaluacion', 'mean')
)
```
